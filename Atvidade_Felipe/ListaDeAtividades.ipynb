```python
def calcular_mediana(valores):

    dados = sorted(valores)

    quantidade = len(dados)

    meio = quantidade // 2

    if quantidade % 2 == 0:
        mediana = (dados[meio - 1] + dados[meio]) / 2
    else:
        mediana = dados[meio]

    return mediana

conjunto_a  = [23,14,31,17,20]
conjunto_b = [75,1,4,3,19,50,64,31]

mediana_a = calcular_mediana(conjunto_a)
print("Mediana do conjunto A:", mediana_a)                                                                                                             
mediana_b = calcular_mediana(conjunto_b)
print("Mediana do conjunto B:", mediana_b)
```

```python
def calcular_moda(lista):
    moda = [1]
    maior_frequencia = 0
    for valor in lista:
        frequencia = lista.count(valor)
        if frequencia > maior_frequencia:
            maior_frequencia = frequencia
            moda = [valor]

        elif frequencia == maior_frequencia and valor not in moda:
            moda.append(valor)

    if maior_frequencia == 1:
        return "Amodal"

    elif len(moda) == 1:
        return "unimodal", moda

    elif len(moda) == 2:
        return "bimodal", moda

    else:
        return "multimodal", moda
    
    return moda

conjunto = [23,14,31,17,23,20,23,31]

dados1 = [10, 20, 20, 30, 40]
dados2 = [10, 10, 20, 20, 30]
dados3 = [10, 10, 20, 20, 30, 30, 40]
dados4 = [10, 20, 30, 40]

print(f"{'='*30}")

moda = calcular_moda(conjunto)
print("Moda do conjunto:", moda)
print(f"{'='*30}")

print(calcular_moda(dados1))
print(calcular_moda(dados2))
print(calcular_moda(dados3))
print(calcular_moda(dados4))
```

```python
def registrar_ocorrencia(codigo, *sistemas, **informacoes):
    print(f"codigo da ccorrencia: {codigo}")
    print(f"quantidade de sistemas afetados: {len(sistemas)}")

    for i, sistema in enumerate(sistemas, start=1): 
        print(f"sistema {i}: {sistema}")

    for chave, valor in informacoes.items():
        print(f"{chave}: {valor}")


registrar_ocorrencia(
    "INC - 001",
    "sistemas",
    "dados"
)
```

```python
def analisar_vendas(nome, *vendas):
    total = 0

    for venda in vendas:
        total += venda

    media = total / len(vendas)
    maior = max(vendas)

    print(f"Vendedor: {nome}")
    print(f"Total vendido: R$ {total:.2f}")
    print(f"Média das vendas: R$ {media:.2f}")
    print(f"Maior venda: R$ {maior:.2f}")
    

    if total >= 5000:
        print("Meta: Atingida")
    else:
        print("Meta: Não atingida")
    print(f"{'='*30}")
```

```python
analisar_vendas(
    "José",
    1200, 800, 1500, 700, 950, 1100, 600, 850
)

analisar_vendas(
    "Pedro",
    1800, 1200, 900, 700, 600
)

analisar_vendas(
    "Lucas",
    500, 1300, 1100, 900
)
```

```python
def analisar_vendas(paciente, *procedimentos, **informacoes):
    print(f"paciente: {paciente}")

    print(f"quantidade de procedimentos: {len(procedimentos)}")

    for i, procedimento in enumerate(procedimentos, start=1):
        print(f"procedimento {i}: {procedimento}")

    print("\ninformações adicionais:")

    for chave, valor in informacoes.items():
        print(f"{chave}: {valor}")

    if "retorno" in informacoes:
        print(f"data de retorno: {informacoes['retorno']}")
    else:
        print("retorno não informado")
```
